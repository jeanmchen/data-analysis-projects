```python
# Analyzing education levels versus aggregated campaign responses

# Aggregating campaign responses

data['total_campaigns_accepted'] = data[['AcceptedCmp1', 'AcceptedCmp2', 'AcceptedCmp3', 'AcceptedCmp4', 'AcceptedCmp5']].sum(axis=1)
data['any_campaign_accepted'] = (data['total_campaigns_accepted'] > 0).astype(int)
data['acceptance_rate'] = data['total_campaigns_accepted'] / 5  

# One-hot encoding education

education_dummies = pd.get_dummies(data['New_Education'], prefix='Education')
data_with_dummies = pd.concat([data, education_dummies], axis=1)

# Calculate correlations between campaign acceptance and education categories

education_columns = education_dummies.columns
var_cor = ['total_campaigns_accepted'] + list(education_columns)
corr_matrix = data_with_dummies[var_cor].corr()

# Plotting 

sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix Heatmap')
plt.show()
```

#### TASK: Explore the relationship between categorical variables and the target variable using bar plots or chi-square tests 

```python
# Hypothesis: Having children affects the campaign response.

from scipy.stats import chi2_contingency

contingency_table = pd.crosstab(data['Kidhome'], data['Response'])

# Chi-square test
chi2, p, dof, expected = chi2_contingency(contingency_table)

print(f"Chi2 Statistic: {chi2}")
print(f"p-value: {p}")
print(f"Degrees of Freedom: {dof}")
print("Expected Frequencies:")
print(expected)
```

<h4 style="color: #cc2222; font-weight: bold;">COMMENT:</h4>

<div style="margin-left: 25px; color: #aa2222;">

The above chi-square test evaluated whether having children has an effect on campaign response.
The chi<sup>2</sup> value is 14.636. This is a very great difference between observed and expected response counts.
The p-value is 0.00066 which indicates an extremely low probability that such a chi<sup>2</sup> value would be observed if there were no effect. Therefore the null hypothesis can be rejected.

Conclusion: Having children has an effect on campaign response.
    
</div>

```python
# Hypothesis: Having teenagers affects the campaign response.

from scipy.stats import chi2_contingency

contingency_table = pd.crosstab(data['Teenhome'], data['Response'])

# Chi-square test
chi2, p, dof, expected = chi2_contingency(contingency_table)

print(f"Chi2 Statistic: {chi2}")
print(f"p-value: {p}")
print(f"Degrees of Freedom: {dof}")
print("Expected Frequencies:")
print(expected)
```

<h4 style="color: #cc2222; font-weight: bold;">COMMENT:</h4>

<div style="margin-left: 25px; color: #aa2222;">

The above chi-square test evaluated whether having teenagers has an effect on campaign response.
